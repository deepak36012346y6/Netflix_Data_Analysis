```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data_df = pd.read_csv('cleaned_dataset.csv')
```

```python
data_df['type'].value_counts()
```

```python
# ceate bar chart
content_counts = data_df['type'].value_counts()
plt.figure(figsize=(8, 5))
sns.barplot(x=content_counts.index, y=content_counts.values)
plt.title('Netflix Content Distribution')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# Calculate content proportions.
content_percentage = (content_counts / content_counts.sum()) * 100
content_percentage
```

```python
"""Key Findings – Netflix Content Type Analysis

1. Movies account for **69.69%** of the total Netflix content.
2. TV Shows account for **30.31%** of the total content.
3. Netflix has significantly more Movies than TV Shows.
4. Movies represent approximately twice the number of TV Shows in the dataset.
Conclusion
The analysis shows that Movies dominate the Netflix content library, representing approximately 70% of the available titles, while TV Shows make up approximately 30%."""
```
