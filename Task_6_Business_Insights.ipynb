```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data_df = pd.read_csv('cleaned_dataset.csv')
```

```python
content_summary = data_df['type'].value_counts().reset_index()
content_summary.columns = ['Content_Type', 'Total_Titles']
content_summary
```

```python
country_df = data_df.assign(
    country=data_df['country'].str.split(',')
).explode('country')
country_df['country'] = country_df['country'].str.strip()
top_countries = country_df[
    country_df['country'] != 'Unknown'
]['country'].value_counts().head(5)
top_countries
```

```python
genre_df = data_df.assign(
    listed_in=data_df['listed_in'].str.split(',')
).explode('listed_in')
genre_df['listed_in'] = genre_df['listed_in'].str.strip()
top_genres = genre_df['listed_in'].value_counts().head(5)
top_genres
```

```python
yearly_content = data_df['release_year'].value_counts()
highest_year = yearly_content.idxmax()
highest_count = yearly_content.max()
print("Year with highest content:", highest_year)
print("Total titles:", highest_count)
```

```python
business_insights = pd.DataFrame({
    'Metric': [
        'Total Netflix Titles',
        'Movies',
        'TV Shows',
        'Top Country',
        'Top Genre',
        'Highest Release Year',
        'Highest Yearly Titles'
    ],
    'Value': [
        len(data_df),
        6126,
        2664,
        'United States',
        'International Movies',
        2018,
        1146
    ]
})

business_insights
```

```python
business_insights.to_csv('netflix_business_insights.csv', index=False)

print("Business Insights Report Saved Successfully!")
```
