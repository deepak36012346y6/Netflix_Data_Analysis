```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data_df = pd.read_csv('cleaned_dataset.csv')
```

```python
rating_counts = data_df['rating'].value_counts()
rating_counts
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

plt.title('Netflix Content Distribution by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
genre_df = data_df.assign(
    listed_in=data_df['listed_in'].str.split(',')
).explode('listed_in')

genre_df['listed_in'] = genre_df['listed_in'].str.strip()

genre_counts = genre_df['listed_in'].value_counts().head(10)

genre_counts
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

```python
genre_report = genre_counts.reset_index()
genre_report.columns = ['Genre', 'Total_Content']
genre_report.to_csv('genre_analysis.csv', index=False)
genre_report
```

### Key Findings – Content Rating & Genre Analysis

- TV-MA is the most common content rating with 3,205 titles.
- TV-14 is the second most common rating with 2,157 titles.
- International Movies is the most popular genre with 2,752 titles.
- Dramas rank second with 2,426 titles.
- Comedies rank third with 1,674 titles.
- Netflix has a diverse collection of movies and TV genres.
