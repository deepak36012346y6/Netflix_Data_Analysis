```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
# Load the dataset
data_df=pd.read_csv(r"C:\Users\dp822\Downloads\Dataset.csv")
data_df
```

```python
data_df.head(),data_df.tail(),data_df.shape
```

```python
data_df.info(),data_df.describe(include="all")
```

```python
# Check missing value
missing_percentage = (data_df.isnull().sum() / len(data_df)) * 100
missing_percentage,data_df.isnull().sum(),
```

```python
data_df['director'] = data_df['director'].fillna('Unknown')
data_df['country'] = data_df['country'].fillna('Unknown')
```

```python
# Check placeholder values
for col in data_df.select_dtypes(include='object').columns:
    print(f'\ncolumn: {col}')
    print(data_df[col].value_counts().head(10))
```

```python
# Handel missing value
data_df.replace('Not Given', np.nan, inplace=True)
data_df.isnull().sum()
```

```python
# Remove duplicates 
data_df.duplicated().sum()
data_df.drop_duplicates(inplace=True)
data_df.columns = data_df.columns.str.strip()
```

```python
# Remove unnecessary spaces from text values
for col in data_df.select_dtypes(include='object').columns:
    data_df[col] = data_df[col].str.strip()
```

```python
data_df.duplicated().sum()
```

```python
# Standardize Country
data_df['country'] = data_df['country'].str.strip()
# Standardize Rating
data_df['rating'] = data_df['rating'].str.strip().str.upper()
data_df['rating'].value_counts()
```

```python
# Convert date_added into datetime
data_df['date_added'] = pd.to_datetime(data_df['date_added'],errors='coerce')
data_df.info()
```

```python
data_df['type'] = data_df['type'].str.title()
data_df['country'] = data_df['country'].str.title()
data_df['rating'] = data_df['rating'].str.upper()
```

```python
# Export the cleaned dataset
data_df.to_csv('cleaned_dataset.csv', index=False)
```
